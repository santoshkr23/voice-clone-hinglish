````
#@title 1. Setup (run once)
!nvidia-smi --query-gpu=name --format=csv,noheader
!git clone -q https://github.com/santoshkr23/voice-clone-hinglish /content/voice 2>/dev/null || git -C /content/voice pull -q
!pip install -q chatterbox-tts==0.1.7 "setuptools<81"
print('Setup done. Now run cell 2.')
````

````
#@title 2. Speak in my voice
text = "Hello everyone, welcome back to my channel. Today I want to share something simple but powerful." #@param {type:"string"}
style = "english" #@param ["calm", "reel", "question", "excited", "soft", "clear", "english", "outro"]
seed = 1 #@param {type:"integer"}
out_name = "my_voice" #@param {type:"string"}
import subprocess, os
open('/content/text.txt', 'w', encoding='utf-8').write(text)
out = f'/content/{out_name}.mp3'
!cd /content/voice && python say.py -f /content/text.txt -s {style} --seed {seed} -o {out}
from IPython.display import Audio, display
display(Audio(out))
print('If a word sounds wrong, change the seed and run again.')
````
